# Decision Trees

## Theory
- Non-parametric supervised learning algorithm
- Used for classification and regression tasks
- Has a hierarchical structure:
  - Root node — represents the entire dataset
  - Internal nodes — represent decisions based on feature values
  - Leaf nodes — represent the predicted output
- Built using a greedy divide-and-conquer strategy
- Pruning helps reduce overfitting by simplifying overly complex trees <br>

## Mathematical theory behind decision trees

### Entropy:
Amount of information need to acurately describe a sample <br>
In a homogenous sample the entropy would be 0, while in a equally divided sample the entropy is 1 <br>
Value between 0 and 1 <br>
$Entropy = -\sum^n_{i=1} p_i * log(p_1)$

```python
import math

def entropy(class_counts):
    total = sum(class_counts)
    ent = 0
    for count in class_counts:
        if count == 0:
            continue
        p = count / total
        ent -= p * math.log2(p)
    return ent


print("Entropy of a pure class (10 positive, 0 negative):",
      entropy([10, 0]))

print("Entropy of evenly split classes (5 positive, 5 negative):",
      entropy([5, 5]))

print("Entropy of imbalanced classes (8 positive, 2 negative):",
      entropy([8, 2]))
```

### Gini index / Gini impurity
A measure of inequality. <br>
Value between 0 and 1 <br>
0 if all sample are perfectly homogenous and all elements are similar <br>
1 describes maximum inequality <br>

$\text{Gini index} = 1 - \sum^n_{i=1} p_i^2$

```python
from sklearn.datasets import load_iris
import pandas as pd

def gini(counts):
    total = sum(counts)
    return 1 - sum((count/total)**2 for count in counts)

iris = load_iris(as_frame=True)
df = iris.frame

subset = df.iloc[:30]

class_counts = subset['target'].value_counts()
print("class_counts:", class_counts)
print("Gini Index:", gini(class_counts))
```

```python
subset2 = pd.concat([df[df.target == 0].iloc[:20],
                     df[df.target == 1].iloc[:20],
                     df[df.target == 2].iloc[:20]])

class_counts2 = subset2['target'].value_counts()
print(class_counts2)
print("Gini Index:", gini(class_counts2))
```

```python
from sklearn import tree
import matplotlib.pyplot as plt
from sklearn.utils import Bunch

def display_decision_tree(model: tree.DecisionTreeClassifier, iris: Bunch) -> None:
    plt.figure(figsize=(12,8))
    tree.plot_tree(model, filled=True, feature_names=iris.feature_names, class_names=iris.target_names)
    plt.show()
```

## Example - Iris Dataset

1. Load the dataset

```python
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
df = iris.frame

df.head()
```

2. Split and train the model

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

X_train, X_test, y_train, y_test = train_test_split(df[iris.feature_names], iris.target, test_size=0.2)

model = DecisionTreeClassifier()
model.fit(X_train, y_train)
```

3. Evaluation of the model

```python
from sklearn.metrics import accuracy_score, classification_report

y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification report:\n", classification_report(y_test, y_pred))
```

4. Visualize the decision tree

```python
display_decision_tree(model, iris)
```

5. Try a different max depth and display the decision tree again

```python
model = DecisionTreeClassifier(max_depth=3)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification report:\n", classification_report(y_test, y_pred))

display_decision_tree(model, iris)
```

## Example from [sci-kit learn](https://scikit-learn.org/1.5/auto_examples/tree/plot_iris_dtc.html)

```python
from sklearn.datasets import load_iris

iris = load_iris()

import matplotlib.pyplot as plt
import numpy as np

from sklearn.datasets import load_iris
from sklearn.inspection import DecisionBoundaryDisplay
from sklearn.tree import DecisionTreeClassifier

# Parameters
n_classes = 3
plot_colors = "ryb"
plot_step = 0.02


for pairidx, pair in enumerate([[0, 1], [0, 2], [0, 3], [1, 2], [1, 3], [2, 3]]):
    # We only take the two corresponding features
    X = iris.data[:, pair]
    y = iris.target

    # Train
    clf = DecisionTreeClassifier().fit(X, y)

    # Plot the decision boundary
    ax = plt.subplot(2, 3, pairidx + 1)
    plt.tight_layout(h_pad=0.5, w_pad=0.5, pad=2.5)
    DecisionBoundaryDisplay.from_estimator(
        clf,
        X,
        cmap=plt.cm.RdYlBu,
        response_method="predict",
        ax=ax,
        xlabel=iris.feature_names[pair[0]],
        ylabel=iris.feature_names[pair[1]],
    )

    # Plot the training points
    for i, color in zip(range(n_classes), plot_colors):
        idx = np.where(y == i)
        plt.scatter(
            X[idx, 0],
            X[idx, 1],
            c=color,
            label=iris.target_names[i],
            edgecolor="black",
            s=15,
        )

plt.suptitle("Decision surface of decision trees trained on pairs of features")
plt.legend(loc="lower right", borderpad=0, handletextpad=0)
_ = plt.axis("tight")
```

## Decision Tree Regressor - Example on the Iris Dataset

Decision Tree Regressor is a supervised learning model that predicts continuous values (numbers) instead of categories.
It splits the input space into regions and assigns each region a mean value of the target variable.

We’ll predict petal length (cm) based on the other flower measurements.

1 - Load the dataset

```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris(as_frame=True)
df = iris.frame

X = df.drop(columns=['petal length (cm)', 'target'])
y = df['petal length (cm)']

df.head()
```

2. Split the dataset and train the model

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

reg = DecisionTreeRegressor(max_depth=3)
reg.fit(X_train, y_train)
```

3. Evaluation of the model

```python
from sklearn.metrics import mean_squared_error, r2_score

y_pred = reg.predict(X_test)

print("Mean Squared Error:", mean_squared_error(y_test, y_pred))
print("R² Score:", r2_score(y_test, y_pred))
```

4. Plot the model

```python
from sklearn import tree
import matplotlib.pyplot as plt

plt.figure(figsize=(14,7))
tree.plot_tree(
    reg,
    filled=True,
    feature_names=X.columns,
    rounded=True,
    fontsize=10
)
plt.title("Decision Tree Regressor to predict Petal Length")
plt.show()
```

6. Feature Importance

```python
feature_importance = pd.Series(reg.feature_importances_, index=X.columns)
feature_importance.sort_values(ascending=False)
```
