# One hot encoding
Models typically cannot be trained on categorical data, so it has to be conveted into numerical data, **one hot encoding** is one of these techniques. <br>
It generates **n** new columns for **n** possible values in the column, e.g. **island** can have 3 possible values ("Dream", "Torgersen", "Biscoe"), therefore it gets three columns. One of these columns is redundant (since if the penguin lives on neither Dream, nor on Torgersen island, it's clear that it lives on  Biscoe island), therefore it can be dropped by using `drop_first=True`.

An alternative to one hot encoding is **label encoding**, in which every category gets replaced with a unique number.

```python
df = pd.get_dummies(df, columns=["island", "sex"], drop_first=True)
df.head()
```
