# One hot encoding
Generates n vectors for n possible values in the column <br>
e.g. embarked -> true / false -> 2 vectors <br>
One with 1 for each row where embarked is true, and 0 otherwise <br>
And the other vector vice versa -> embarked = false -> 1, else 0 <br>
The second vector is redundant, therefore we drop it `drop_first=True`

```python
df = pd.get_dummies(df, columns=['embarked', 'sex', 'who', 'adult_male', 'embark_town'], drop_first=True)
df.head()
```
